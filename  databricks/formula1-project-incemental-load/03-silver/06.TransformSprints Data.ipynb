```python
dbutils.widgets.text("p_batch_id", "")
v_batch_id = dbutils.widgets.get("p_batch_id")
```

```python
bronze_table= f"{catalog_name}.{bronze_schema}.sprints"
silver_table= f"{catalog_name}.{silver_schema}.sprints"
```

```python
sprints_df = (spark.read.table(bronze_table)
              .drop("url")
              .withColumnsRenamed({ "constructorId": "constructor_id", 
                                   "driverId": "driver_id",
                                   "raceName": "race_name",
                                   "positionText": "final_position_text",
                                   "date": "race_date",
                                   "grid": "grid_position",
                                   "laps": "completed_laps",
                                   "number": "car_number",
                                   "position": "final_position"})

)
```

```python
from pyspark.sql import functions as F
```

```python
sprints_fil_df= (sprints_df
                 .filter(F.col("season").isNotNull() & F.col("round").isNotNull() & F.col("constructor_id").isNotNull() & F.col("driver_id").isNotNull())
                 .dropDuplicates(["season", "round", "constructor_id","driver_id"]))
```

```python
display(sprints_df.count() - sprints_fil_df.count())
```

```python
sprints_final_df= (sprints_fil_df.withColumn("race_name", F.initcap(F.col("race_name")))
 .withColumn("batch_id", F.lit(v_batch_id))                  
)
```

```python
display(sprints_final_df)
```

```python
write_to_silver(
    input_df= sprints_final_df,
    target_table= silver_table,
    merge_condition= "t.season=s.season AND t.round=s.round AND t.constructor_id= s.constructor_id AND t.driver_id= s.driver_id",
    columns_to_update=[
        "race_name",
        "race_date",
        "grid_position",
        "completed_laps",
        "car_number",
        "points",
        "final_position",
        "final_position_text",
        "status",
        "ingestion_timestamp",
        "source_file",
        "batch_id"
    ]
)
```
