## Understanding Delta lkae Transaction Log
Understanding the cloud storage directory structure for delta table

#### 1. Create Catalog and Schema for the Demo

**Catalog:** demo 
**Schema:** delta_lake

```sql
CREATE CATALOG IF NOT EXISTS demo
MANAGED LOCATION 'abfss://demo@brickscourseextdl.dfs.core.windows.net/'
```

```sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
MANAGED LOCATION 'abfss://demo@brickscourseextdl.dfs.core.windows.net/delta_lake'
```

## CREATE DELTA LAKE TABLE 

```sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
(
    company_name STRING,
    founded_date DATE,
    country STRING
) 
USING DELTA;
```

```sql
DESC EXTENDED demo.delta_lake.companies
```
