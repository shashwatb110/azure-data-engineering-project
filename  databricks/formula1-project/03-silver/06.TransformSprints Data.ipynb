```python
bronze_table= f"{catalog_name}.{bronze_schema}.sprints"
silver_table= f"{catalog_name}.{silver_schema}.sprints"
```

```python
sprints_df = (spark.read.table(bronze_table)
              .drop("url")
              .withColumnsRenamed({ "constructorId": "constructor_id", 
                                   "driverId": "driver_id",
                                   "raceName": "race_name",
                                   "positionText": "final_position_text",
                                   "date": "race_date",
                                   "grid": "grid_position",
                                   "laps": "completed_laps",
                                   "number": "car_number",
                                   "position": "final_position"})

)
```

```python
from pyspark.sql import functions as F
```

```python
sprints_fil_df= (sprints_df
                 .filter(F.col("season").isNotNull() & F.col("round").isNotNull() & F.col("constructor_id").isNotNull() & F.col("driver_id").isNotNull())
                 .dropDuplicates(["season", "round", "constructor_id","driver_id"]))
```

```python
display(sprints_df.count() - sprints_fil_df.count())
```

```python
sprints_final_df= sprints_fil_df.withColumn("race_name", F.initcap(F.col("race_name")))
```

```python
(
    sprints_final_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(silver_table)
)
```
