```python
source_file = f'{landing_folder_path}/constructors.json'
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

## Read JSON file using the dataframe reader API

```python
# DEFINE SCHEMA
constructors_schema= "constructorId String, name String, nationality String, url String"
```

```python
constructor_df= (
    spark.read
    .format('json')
    .schema(constructors_schema)
    .option('mode', 'FAILFAST')
    .load(source_file)
)
```

```python
display(constructor_df)
```

## ADDING METADATA

```python
constructor_final_df= add_ingestion_metadata(constructor_df)
```

```python
display(constructor_final_df)
```

```python
(
    constructor_final_df
.write
.format('delta')
.mode('overwrite')
.saveAsTable(table_name)
)
```
